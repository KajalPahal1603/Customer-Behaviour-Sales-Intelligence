```python
import pandas as pd
import numpy as np

print("Pandas version:", pd.__version__)
print("NumPy version:", np.__version__)
print("Project started successfully!")
```

```python
df = pd.read_excel("../data/Online Retail.xlsx")

print("Dataset loaded successfully!")
print("Rows and columns:", df.shape)

df.head()
```

```python
df.info()
```

```python
# Count missing values in each column
missing_values = df.isnull().sum()

print("Missing values in each column:")
print(missing_values)
```

```python
# Calculate missing value percentages
missing_percentage = (df.isnull().sum() / len(df)) * 100

print("Missing values percentage:")
print(missing_percentage.round(2))
```

```python
# Check duplicate records
duplicate_count = df.duplicated().sum()

print("Total duplicate records:", duplicate_count)
```

```python
# Check cancelled transactions
cancelled_count = df["InvoiceNo"].astype(str).str.startswith("C").sum()

print("Total cancelled transactions:", cancelled_count)
```

```python
# Check negative quantities
negative_quantity_count = (df["Quantity"] < 0).sum()

print("Total negative quantity records:", negative_quantity_count)
```

```python
# Check zero and negative unit prices
invalid_price_count = (df["UnitPrice"] <= 0).sum()

print("Total zero or negative price records:", invalid_price_count)
```

```python
# Check transaction date range
print("First transaction date:", df["InvoiceDate"].min())
print("Last transaction date:", df["InvoiceDate"].max())
```

```python
# Check top 10 countries by transaction rows
top_countries = df["Country"].value_counts().head(10)

print("Top 10 countries by transaction rows:")
print(top_countries)
```

```python
# Check total unique customers
unique_customers = df["CustomerID"].nunique()

print("Total unique customers:", unique_customers)
```

```python
# Check total unique products
unique_products = df["StockCode"].nunique()

print("Total unique products:", unique_products)
```

```python
# Check total unique invoices
unique_invoices = df["InvoiceNo"].nunique()

print("Total unique invoices:", unique_invoices)
```
